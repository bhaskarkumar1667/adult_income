```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('adult_cleaned.csv')
```

```python
x=df.drop('class',axis=1)
y=df['class']
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import AdaBoostClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
x_test, x_train, y_test, y_train = train_test_split(x, y, test_size=0.2, random_state=42)
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
knn_param={'n_neighbors':[3,5,7,9,11,13,15],'weights':['uniform','distance'],'algorithm':['auto','ball_tree','kd_tree','brute']}
lr_param={'penalty':['l1','l2','elasticnet'],'C':[0.01,0.1,1,10,100],'solver':['newton-cg','lbfgs','liblinear','sag','saga']}
svm_param={'C':[0.01,0.1,1,10,100],'kernel':['linear','poly','rbf','sigmoid'],'gamma':['scale','auto']}
dt_param={'criterion':['gini','entropy'],'max_depth':[None,5,10,15,20],'min_samples_split':[2,5,10],'min_samples_leaf':[1,2,4]}
gb_param={'n_estimators':[50,100,200],'learning_rate':[0.1,0.01,0.001],'max_depth':[3,5,7]}
rf_param={'n_estimators':[50,100,200],'max_depth':[None,5,10,15],'min_samples_split':[2,5,10],'min_samples_leaf':[1,2,4]}
xgb_param={'n_estimators':[50,100,200],'learning_rate':[0.1,0.01,0.001],'max_depth':[3,5,7]}
ad_param={'n_estimators':[50,100,200],'learning_rate':[0.1,0.01,0.001]}
grid_params={'knn':knn_param,'lr':lr_param,'svm':svm_param,'dt':dt_param,'gb':gb_param,'rf':rf_param,'xgb':xgb_param,'ad':ad_param}
models={'knn':KNeighborsClassifier(),'lr':LogisticRegression(),'svm':SVC(),'dt':DecisionTreeClassifier(),'gb':GradientBoostingClassifier(),'rf':RandomForestClassifier(),'xgb':XGBClassifier(),'ad':AdaBoostClassifier()}
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
i=0
for name,params in grid_params.items():
    model=models[name]
    i+=1
    gr=RandomizedSearchCV(model,params,cv=5,n_iter=10,scoring='accuracy')
    gr.fit(x_train,y_train)
    gr_pred=gr.predict(x_test)
    print(f"Model: {model}")
    print(f"Best Parameters: {gr.best_params_}")
    print(f"Accuracy: {accuracy_score(y_test,gr_pred)}")    
    print(f"Precision: {precision_score(y_test,gr_pred)}")
    print(f"Recall: {recall_score(y_test,gr_pred)}")
    print(f"F1 Score: {f1_score(y_test,gr_pred)}")
    print(f"ROC AUC Score: {roc_auc_score(y_test,gr_pred)}")
```
